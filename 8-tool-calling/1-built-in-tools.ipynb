# Tool Calling

LLMs cannot reliably answer questions about events that happened after their knowledge cutoff. Let's first see this limitation and then give the model access to current information.

```python
from dotenv import load_dotenv
from openai import OpenAI


load_dotenv(override=True)

client = OpenAI()
model = "gpt-6-luna"
prompt = "What big news regarding AI Engineering happened in the last 7 days?"
```

## The problem

Without access to current information, the model cannot reliably answer our question.

```python
response = client.responses.create(
    model=model,
    input=prompt,
)

print(response.output_text)
```

```python
# Give the model access to current information
response = client.responses.create(
    model=model,
    tools=[{"type": "web_search"}],
    input=prompt,
)

print(response.output_text)
```

OpenAI's built-in web search tool searches for current information before the model answers the question.

**Cost:** Using web search makes the request more expensive. In addition to the model's regular input and output token costs, each web search has a tool-call fee, and the retrieved search content is billed at the model's input-token rate. See the [OpenAI API pricing](https://developers.openai.com/api/docs/pricing).
